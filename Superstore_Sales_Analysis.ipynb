```python
# ==========================================
# Task 2: Superstore Sales Data Analysis
# Internship: SYNENT Technologies
# Author: Unnati Gore
# ==========================================

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Style for Visuals
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

# 1. Sample Dataset Creation (Superstore Sales)
data = {
    'Order_ID': [f'ORD-{100+i}' for i in range(20)],
    'Category': ['Furniture', 'Technology', 'Office Supplies', 'Furniture', 'Technology',
                 'Office Supplies', 'Furniture', 'Technology', 'Office Supplies', 'Furniture']*2,
    'Sub_Category': ['Chairs', 'Phones', 'Paper', 'Tables', 'Accessories',
                     'Binders', 'Furnishings', 'Copiers', 'Appliances', 'Bookcases']*2,
    'Sales': [250, 800, 45, 450, 150, 20, 120, 1200, 300, 600,
              210, 950, 60, 500, 180, 35, 140, 1100, 280, 650],
    'Profit': [40, 180, 12, 50, 30, 5, 25, 300, 60, 90,
               30, 220, 15, 60, 40, 8, 30, 280, 50, 100],
    'Region': ['East', 'West', 'Central', 'South', 'East', 'West', 'Central', 'South', 'East', 'West']*2
}

df = pd.DataFrame(data)

print("--- Dataset Head ---")
print(df.head())

# 2. Total Sales and Profit by Category
category_summary = df.groupby('Category')[['Sales', 'Profit']].sum().reset_index()

# Visualization 1: Category wise Sales & Profit
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=category_summary, x='Category', y='Sales', hue='Category', ax=ax[0], palette='Blues_d', legend=False)
ax[0].set_title('Total Sales by Category', fontsize=14, fontweight='bold')
ax[0].set_ylabel('Sales ($)')

sns.barplot(data=category_summary, x='Category', y='Profit', hue='Category', ax=ax[1], palette='Greens_d', legend=False)
ax[1].set_title('Total Profit by Category', fontsize=14, fontweight='bold')
ax[1].set_ylabel('Profit ($)')

plt.tight_layout()
plt.savefig('category_sales_profit.png')
plt.show()

# Visualization 2: Top Sub-Categories by Sales
plt.figure(figsize=(10, 5))
sns.barplot(data=df.sort_values(by='Sales', ascending=False), x='Sales', y='Sub_Category', hue='Sub_Category', palette='magma', legend=False)
plt.title('Top Sub-Categories by Total Sales', fontsize=14, fontweight='bold')
plt.xlabel('Sales ($)')
plt.ylabel('Sub-Category')
plt.savefig('sub_category_sales.png')
plt.show()

print("\nTask 2 Analysis Completed Successfully!")
```
